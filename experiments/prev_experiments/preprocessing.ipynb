```python
n_exist_cnt = 0
for s in df_seq_data.seq:
    flag = False
    for n in s:
        if n not in ('ATCG'):
            flag = True
            # print('hit:', n)
    if flag:
        # print(s)
        n_exist_cnt += 1
print(n_exist_cnt) # exist 5 transcript with N nucleotide, whether to drog these 5 transcript?
```

```python
df_data = pd.DataFrame()

df_data['label'] = (df_label.characteristics == 'positive') + 0
df_data['position'] = df_label.py_position
df_data['seq'] = df_label.apply(lambda row: df_seq_data.loc[row.gene_ID, 'seq'],
                                 axis=1)
```
